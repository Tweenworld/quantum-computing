### Why `np.vdot` Instead of `np.dot`?

In quantum mechanics, state vectors exist in a complex vector space. The inner product of two states $|\psi\rangle$ and $|\phi\rangle$ is defined as the bracket $\langle \psi | \phi \rangle$. This operation requires taking the **conjugate transpose** (Hermitian conjugate, $\dagger$) of the first vector:

$$\langle \psi | \phi \rangle = \psi^\dagger \phi = \sum_{k} \psi_k^* \phi_k$$

* **`np.vdot(A, B)`**: Specially designed for complex vectors. It automatically takes the **complex conjugate of the first argument** (`A`) before computing the dot product.
* **`np.dot(A, B)`**: Performs a standard matrix multiplication without conjugation.

*If you run `np.dot(v1.T, v2)` on these real vectors, you will get the same result, but it will fail to yield the correct physical inner product once complex numbers are introduced.*

## 2. Quantum States: Complex Vector Inner Product

Complex coefficients. Consider the following two state vectors:

$$\mathbf{v}_1 = \begin{pmatrix} i \\ 1 \end{pmatrix}, \quad \mathbf{v}_2 = \begin{pmatrix} -i \\ 1 \end{pmatrix}$$

### Calculation Challenge

Before writing your code, compute the inner product $\langle \mathbf{v}_1 | \mathbf{v}_2 \rangle$ manually. Remember to conjugate the elements of the first vector ($\mathbf{v}_1$):

$$\langle \mathbf{v}_1 | \mathbf{v}_2 \rangle = \mathbf{v}_1^\dagger \mathbf{v}_2 = \begin{pmatrix} -i & 1 \end{pmatrix} \begin{pmatrix} -i \\ 1 \end{pmatrix} = (-i)(-i) + (1)(1) = i^2 + 1 = -1 + 1 = 0$$

Because their inner product is $0$, these two state vectors are **orthogonal** (representing physically independent quantum states).


```python
import numpy as np

# Define v1 and v2 as column vectors (2x1 matrices)
v1 = np.array([[np.sqrt(3)], [1]])
v2 = np.array([[1], [1]])

# Complex inner product (conjugate transpose of the first vector)
inner_result = np.vdot(v1, v2)

print(f"v1:\n{v1}\n")
print(f"v2:\n{v2}\n")
print(f"Inner Product <v1|v2>: {inner_result}")
```

```python
# Define vectors with complex coefficients
v1 = np.array([[1j], [1]])
v2 = np.array([[-1j], [1]])

# Compute the quantum inner product
inner_result = np.vdot(v1, v2)

print(f"v1:\n{v1}\n")
print(f"v2:\n{v2}\n")
print(f"Quantum Inner Product <v1|v2>: {inner_result}")
```
